```python
import ujson as json
```

```python
PATH = "../../web/all_cards.json"
OUTPUT_PATH = "../../web/cards.json"

cards = json.load(open(PATH, "r"))
```

```python
from collections import Counter
from random import Random

r = Random()
r.seed(513)

by_id = {}
for c in cards:
    if not c["study"]["nct_id"] in by_id:
        by_id[c["study"]["nct_id"]] = [c]
    else:
        by_id[c["study"]["nct_id"]].append(c)

for id, cs in by_id.items():
    if len(cs) > 3:
        by_id[id] = r.sample(cs, 3)

final = []
for cs in by_id.values():
    final.extend(cs)

json.dump(final, open(OUTPUT_PATH, "w"))
```
